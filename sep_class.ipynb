```python
#recurssion : a function calling itself

#wap to find factorial of a number using recurrsion

n=int(input("Enter the number : "))
def fac(n):
  if n==1:
    return 1
  else:
    return n*fac(n-1)

fac(n)
```

```python
#wap to find the sum of n natural number using recurssion

n=int(input("Enter the number : "))
def sum(n):
  if n==1:
    return 1
  else:
    return n+sum(n-1)

sum(n)
```

```python
#create a class of circles and print area and perimeter using methods

class Circles():
  def get_data(self):
    self.radius=int(input("Enter radius : "))

  def display_data(self):
    print(self.radius)

  def area(self):
    print(3.14*(self.radius**2))
  def perimeter(self):
    print(2*3.14*self.radius)

c1=Circles()
c1.get_data()
c1.area()
c1.perimeter()
c2=Circles()
c2.get_data()
c2.area()
c2.perimeter()
```

```python
#OOP CONCEPTS

#1. Abstraction: Act of representing essential features by hiding the background details
  #Access modifiers:used to control the access of data members in a class
   #public : can access everywhere
   #protected: single underscore before the attribute
   #private: double underscore before the attribute

#2. Encapsulation: Binding of data members and methods together
#3. Inheritance: Passing of properties of parent class to child class
   #1.single inheritance
   #2.Multilevel inheritance: imagine grandfather and father (son can access both of them)
   #3 Multiple inheritance: imagine father(son and daughter can access both of them)
   #4 Hierarchical inheritance: imagine father and mother (son has the properties of both of them)
   #5 Hybrid inheritance: combination of any of the above inheritance method

#4. Polymorphism: Ability to exist in many forms

   #1. Method overloading: call the function according to method(not supported in python)
   #2. OPERATOR OVERLOADING: Sme operator used for different situations(eg: 1+2=3,"1"+"2"="12")
   #3. METHOD OVERRIDING: PARENT AND CHILD CLASS HAS THE SAME METHOD,CHILD CLASS OBJECT IS OVERRIDED BY PARENT CLASS
   #4. FUNCTIONAL POLYMORPHISM: eg:len("appple") and len([1,2,2,3])

#5. Duck typing: 2 object has the same function and called according to the object
```

```python
class Addition():
  def __init__(self,n):
    self.value=n                                      #Magic methods/Magic functions
  def __add__(self,other):                            #__add__()
    return self.value+other.value                     #__sub__()
                                                      #__str__()

a1=Addition(10)
a2=Addition(15)

print(a1+a2)
```
